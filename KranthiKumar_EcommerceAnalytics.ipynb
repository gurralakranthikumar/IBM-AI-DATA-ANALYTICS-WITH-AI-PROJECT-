```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

print("1. Loading datasets...")
df_sales = pd.read_csv('ecommerce_sales_customer_analytics_150k.csv')
df_customers = pd.read_csv('customer_master.csv')
df_products = pd.read_csv('product_catalog.csv')
df_orders = pd.read_csv('order_items.csv')

print("\n2. Basic Data Overview:")
print(df_sales.info())

print("\n3. Preparing Data for AI Model...")
# Select numeric columns for the AI model and drop missing values
numeric_df = df_sales.select_dtypes(include=[np.number]).dropna()

# Standardize the data so the algorithm processes all values equally
scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_df)

print("\n4. Running K-Means Clustering (AI Segmentation)...")
# Apply K-Means to segment the data into 3 distinct customer behavior profiles
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
numeric_df['AI_Customer_Segment'] = kmeans.fit_predict(scaled_data)

print("\nSuccess! AI Segmentation Complete. Here is a sample of the segmented output:")
print(numeric_df[['AI_Customer_Segment']].head())
print("\nYou can now go to File > Download > Download .ipynb")
```
